```python
masque = (df["Sales"] - df["Sales"].mean()).abs() > (
    3 * df["Sales"].std()
)
df[masque]
masque.sum()
```

```python
df["Product_general_price"] = (df["Sales"]/df["Quantity"])+(df["Sales"]*df["Discount"])/(df["Quantity"])
outliers = df.groupby("Product ID")["Product_general_price"].agg(
    Q1 = lambda sale: sale.quantile(0.25),
    Q3 = lambda sale: sale.quantile(0.75)
)
outliers["IQR"]=outliers["Q3"]-outliers["Q1"]
outliers["borne_inf"] = outliers["Q1"] - 1.5 * outliers["IQR"]
outliers["borne_sup"] = outliers["Q3"] + 1.5 * outliers["IQR"]
df = df.join(outliers[["borne_inf", "borne_sup"]], on="Product ID")
masque = (
    (df["Product_general_price"] < df["borne_inf"])|(df["Product_general_price"] > df["borne_sup"])
)
masque.sum()
df[masque]
```
